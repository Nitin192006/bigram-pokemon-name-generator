```python
import torch
```

```python
words = open('pokemon.txt', 'r').read().splitlines()    #reading data
```

```python
N = torch.zeros((27, 27), dtype=torch.int32)    #initializing bigram table
```

```python
#mapping characters and indexes ('.' -> 0)

chars = sorted(list(set(''.join(words))))
stoi = {s:i+1 for i,s in enumerate(chars)}
stoi['.'] = 0
itos = {i:s for s,i in stoi.items()}
```

```python
#forming bigram pairs and storing their count

for w in words:
  chs = ['.'] + list(w) + ['.']
  for ch1, ch2 in zip(chs, chs[1:]):
    ix1 = stoi[ch1]
    ix2 = stoi[ch2]
    N[ix1, ix2] += 1
```

```python
#visualising bigram table

import matplotlib.pyplot as plt
%matplotlib inline
plt.style.use('seaborn-v0_8-dark')
plt.figure(figsize=(18,18), dpi=160)
plt.imshow(N, cmap='turbo', interpolation='nearest')
plt.colorbar(fraction=0.046, pad=0.04)
for i in range(27):
    for j in range(27):
        chstr = itos[i] + itos[j]
        plt.text(j, i, chstr,ha="center", va="bottom",color='white', fontsize=9, alpha=0.9)
        plt.text(j, i, N[i, j].item(),ha="center", va="top",color='white', fontsize=9, fontweight='bold')
plt.axis('off')
plt.tight_layout()
```

```python
#converting count to probability

P=(N+1).float()
P=P/P.sum(1,keepdims=True)    #sums and divides along the rows
```

```python
import matplotlib.pyplot as plt
letters = ['a', 'e', 'i', 'o', 'u']
plt.style.use('seaborn-v0_8-dark')
fig, axes = plt.subplots(len(letters),1,figsize=(18, 24),dpi=200)

for ax, ch in zip(axes, letters):
    idx = stoi[ch]
    probs = P[idx]
    values, indices = torch.topk(probs, 8)
    chars = [itos[int(i)] for i in indices]
    vals = values.numpy()

    bars = ax.bar(chars,vals,width=0.35)
    # value labels
    for bar, v in zip(bars, vals):
        ax.text(bar.get_x() + bar.get_width()/2,bar.get_height(),f"{v:.3f}",ha='center',va='bottom',fontsize=12,fontweight='bold')
        ax.set_title(f"Most Likely Next Letters After '{ch}'",fontsize=22,fontweight='bold',pad=15)
        ax.set_ylabel("Probability",fontsize=15)
        ax.tick_params(axis='x',labelsize=15)
        ax.tick_params(axis='y',labelsize=12)
        ax.grid(alpha=0.25,linestyle='--')
plt.suptitle(
    "Learned Character Transition Probabilities",fontsize=30,fontweight='bold',y=1.02
)
plt.tight_layout()
plt.savefig(
    "character_transitions.png",
    dpi=300,
    bbox_inches='tight'
)
plt.show()
```

```python
#predicting next char in the pair using P

g = torch.Generator().manual_seed(2147483647)
for i in range(10):
  out = []
  ix = 0
  while True:
    p = P[ix]
    ix = torch.multinomial(p, num_samples=1, replacement=True, generator=g).item()
    out.append(itos[ix])
    if ix == 0:
      break
  print(''.join(out))
```

```python
#calculatinh maximum likelihood

log_likelihood = 0.0
n = 0

for w in words:
  chs = ['.'] + list(w) + ['.']
  for ch1, ch2 in zip(chs, chs[1:]):
    ix1 = stoi[ch1]
    ix2 = stoi[ch2]
    prob = P[ix1, ix2]
    logprob = torch.log(prob)
    log_likelihood += logprob
    n += 1
print(f'{log_likelihood=}')
nll = -log_likelihood
print(f'{nll=}')
print(f'{nll/n}')
```

```python
#NN MODEL:
```

```python
import torch.nn.functional as F
```

```python
#Model input for one word

xs,ys=[],[]

for w in words[:1]:
  chs = ['.'] + list(w) + ['.']
  for ch1, ch2 in zip(chs, chs[1:]):
      ix1 = stoi[ch1]
      ix2 = stoi[ch2]
      xs.append(ix1)
      ys.append(ix2)

xs=torch.tensor(xs)
ys=torch.tensor(ys)
num = xs.nelement()
print('number of examples: ', num)
xenc = F.one_hot(xs, num_classes=27).float()
plt.imshow(xenc)
```

```python
#creating input and labels for input layer

xs,ys=[],[]

for w in words:
  chs = ['.'] + list(w) + ['.']
  for ch1, ch2 in zip(chs, chs[1:]):
      ix1 = stoi[ch1]
      ix2 = stoi[ch2]
      xs.append(ix1)
      ys.append(ix2)

xs=torch.tensor(xs)
ys=torch.tensor(ys)
num = xs.nelement()
print('number of examples: ', num)
```

```python
g = torch.Generator().manual_seed(2147483647)
W=torch.randn((27,27),generator=g,requires_grad=True)    #generating random weights
```

```python
#one hot encoding the inputs
xenc = F.one_hot(xs, num_classes=27).float()
```

```python
#gradient descent

for k in range(50):
  logits = xenc @ W 
  counts = logits.exp()
  probs = counts / counts.sum(1, keepdims=True)    #generating probability according to the encoded input and NN weights
  loss = -probs[torch.arange(num), ys].log().mean() + 0.01*(W**2).mean()    #getting probability values of actual labels
  print(loss.item())
  W.grad = None    #resetting wieghts grad
  loss.backward()    #back propogation on loss to tune the weights
  W.data += -100* W.grad    #changing W values according to W.grad(which is w.r.t loss)
```

```python
plt.style.use('seaborn-v0_8-dark')
W_vis = W.detach().numpy()
plt.figure(figsize=(18,18), dpi=160)
plt.imshow(W_vis, cmap='turbo', interpolation='nearest')
plt.colorbar(fraction=0.046, pad=0.04)
for i in range(27):
    for j in range(27):
        chstr = itos[i] + itos[j]
        plt.text(
            j,
            i,
            chstr,
            ha="center",
            va="bottom",
            color="white",
            fontsize=9,
            alpha=0.8
        )
        plt.text(
            j,
            i,
            f"{W_vis[i,j]:.1f}",
            ha="center",
            va="top",
            color="white",
fontsize=8,fontweight="bold"
        )
plt.title(
    "Neural Network Learned Weights",
    fontsize=24,
    pad=20,
    fontweight='bold'
)
plt.axis('off')
plt.tight_layout()
plt.savefig(
    "nn_weights.png",
    dpi=300,
    bbox_inches='tight'
)
plt.show()
```

```python
#Trying to find the new Pokemon. Gotta catch'em all!

g = torch.Generator().manual_seed(2147483647)
for i in range(5):
  out = []
  ix = 0
  while True:
    xenc = F.one_hot(torch.tensor([ix]), num_classes=27).float()
    logits = xenc @ W # predict log-counts
    counts = logits.exp() # counts, equivalent to N
    p = counts / counts.sum(1, keepdims=True)
    ix = torch.multinomial(p, num_samples=1, replacement=True, generator=g).item()
    out.append(itos[ix])
    if ix == 0:
      break
  print(''.join(out))
```
